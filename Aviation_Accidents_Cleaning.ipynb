```python
#Convert Dates to date data type and then filter df to only filter from the year 1983
df["Event.Date"] = pd.to_datetime(df["Event.Date"], errors="coerce")
#Creating a year column
df["Year"] = df["Event.Date"].dt.year
#Filter dataset (1983 onwards ONLY)
df = df[df["Year"] >= 1983]
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# removing missing values first.
df["Total.Fatal.Injuries"] = df["Total.Fatal.Injuries"].fillna(0)
df["Total.Serious.Injuries"] = df["Total.Serious.Injuries"].fillna(0)
df["Total.Minor.Injuries"] = df["Total.Minor.Injuries"].fillna(0)
df["Total.Uninjured"] = df["Total.Uninjured"].fillna(0)
```

```python
# Estimating total passengers onboard (Total On Board=Fatal+Serious+Minor+Uninjured)
df["Total.On.Board"] = (
    df["Total.Fatal.Injuries"] +
    df["Total.Serious.Injuries"] +
    df["Total.Minor.Injuries"] +
    df["Total.Uninjured"]
)
```

```python
# Removing invalid flights (Total.On.Board < 0)
df = df[df["Total.On.Board"] > 0]
```

```python
# Constructing Fatal Injury likelihood metric(Fatal Injury Likelihood= Total Fatal Injuries/Total On Board )
df['Fatal.Injury.Likelihood'] = (
    df['Total.Fatal.Injuries'] / df['Total.On.Board']
)
```

```python
# Construct Serious Injury likelihood metric (Serious Injury Likelihood=Total Serious Injuries/Total On Board)

df["Serious.Injury.Likelihood"] = (
    df["Total.Serious.Injuries"] / df["Total.On.Board"]
)
```

```python
# Combined risk score Total Injury Risk=  (Fatal + Serious Injuries)/Total On Board
df["Severe.Injury.Risk"] = (
    df["Total.Fatal.Injuries"] +
    df["Total.Serious.Injuries"]
) / df["Total.On.Board"]
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
#inspecting values 
df["Aircraft.damage"].value_counts(dropna=False)
```

```python
# standardize text formart for consistency
df["Aircraft.damage"] = df["Aircraft.damage"].str.lower().str.strip()
```

```python
# Create “Destroyed Aircraft” column by converting damage into a binary safety metric. 1 = aircraft destroyed and 0 = not destroyed
df["Aircraft.Destroyed"] = df["Aircraft.damage"].apply(
    lambda x: 1 if x == "destroyed" else 0
)
```
